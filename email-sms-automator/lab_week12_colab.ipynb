## 5. ทดลองค้นหาอีเมลด้วย IMAP โดยตรง

```python
import sys
sys.path.append('src')
from email_receiver import EmailReceiver
r = EmailReceiver('imap.gmail.com', sender, userdata.get('SENDER_APP_PASSWORD').replace(' ', ''))
uids = r.search_emails(criteria='ALL', subject_contains='Python Test')
print('พบอีเมล', len(uids), 'ฉบับ')
for uid in uids[-3:]:
    d = r.fetch_email_content(uid, mark_as_read=False)
    print('-', d['subject'], '|', d['body_snippet'].splitlines()[0])
```
